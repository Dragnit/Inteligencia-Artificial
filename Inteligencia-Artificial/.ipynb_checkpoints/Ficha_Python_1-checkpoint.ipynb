```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# 1 ler o CSV
df = pd.read_csv("Dataset_python.csv")
```

```python
# 2 ver as 15 primeiras linhas
print(df.head(15))
```

```python
# 3 Obter a dimensão do dataset
print(df.shape)
```

```python
# 4 remover a primeira linha do dataset
df.to_csv("Dataset_python.csv", index=False, header=False)
print(df.head(2))
```

```python
# 5 adicionar headers novos
n = df.shape[1]
df.columns = [f"coluna{i+1}" for i in range(n)]
df.to_csv("Dataset_python.csv", index=False)
print(df.head(2))
```

```python
# 6 apresentar o nome das colunas
print(df.columns)
```

```python
# 7 trocar o nome de uma coluna a escolha
df.rename(columns={'coluna1': 'Nome'}, inplace=True)
print(df.head(2))
```

```python
# 8 trocar o nome da primeira coluna de index para indice
df.index.name = 'Indice'
print(df.head())
```

```python
#  9 Obter um sumário da estatística descritiva para cada coluna
summary = df.describe(include="all")
print(summary)
```

```python
# 10 obter um sumário da estatística descritiva somente para variáveis categóricas.
summary = df.describe(include=['object'])
print(summary)
```

```python
# 11 obter um sumário da estatística descritiva somente para uma coluna (à escolha), 
# respeitante a uma variável de escala
summary = df[['coluna3', 'coluna2']].describe()
print(summary)
```

```python
# 12 obter os valores máximos e mínimos da coluna anterior.
max = df['coluna2'].max()
min = df['coluna2'].min()
print(max)
print(min)
```

```python
# 13 A presentaremos a frequência de cada um dos valores possíveis de uma coluna (à escolha), 
# respeitante a uma variável categórica
frequency = df['coluna2'].value_counts()
print(frequency)
```

```python
# 14 criar um dataframe contendo cada um dos valores possíveis da coluna da alínea anterior.
df['coluna2'].value_counts().to_frame()
```

```python
# 15 Apresentar graficamente o histograma de uma coluna à escolha.
plt.hist(df["coluna2"])
plt.xlabel("column")
plt.ylabel("count")
plt.title("Histogram")
plt.show()
```

```python
# 16 apresentar as categorias existentes na coluna2, respeitante a uma variável categórica.
df['coluna2'].unique()
```

```python
# 17 apresentar em boxplot a relação entre as clunas 1 e 2 
df_filtrado = df.head(10)  # só 10 categorias
sns.boxplot(x="coluna1", y="coluna2", data=df_filtrado)
plt.xticks(rotation=90)
plt.show()

df["grupo"] = pd.qcut(df["coluna2"], 4, labels=["baixo","medio","alto","muito_alto"])

sns.boxplot(x="grupo", y="coluna2", data=df)
plt.show()
```

```python
# 18 Substituir numa coluna respeitante a uma variável de escala,
# um determinado valor por NaN.
df["coluna2"] = df["coluna2"].replace(10050, np.nan)
```

```python
# 19 obter o número de "missing values" presentes na coluna em causa.
missing_data = df.isnull()
print(missing_data["coluna2"].value_counts())
```

```python
# 20 substituir os “missing values” presentes na coluna em causa, pela média da mesma.
media = df["coluna2"].mean()
df["coluna2"] = df["coluna2"].replace(np.nan, media)
```

```python
# 22 eliminar os “missing values” da coluna em causa.
df.dropna(subset=["coluna2"], axis=0, inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
# 24 obtero tipo de dados de cada coluna
df.dtypes, df.info()
```

```python
# 25 converter a coluna34 para datetime
df["coluna34"] = pd.to_datetime(df["coluna34"], errors="coerce")
print(df["coluna34"].dtype)
```

```python
# 26
df["coluna2_norm"] = df["coluna2"] / df["coluna2"].max()
print(df["coluna2_norm"].head())
```

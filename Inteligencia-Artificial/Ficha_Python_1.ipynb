```python
# 17 apresentar em boxplot a relação entre as clunas 1 e 2 
df_filtrado = df.head(10)  # só 10 categorias
sns.boxplot(x="coluna1", y="coluna2", data=df_filtrado)
plt.xticks(rotation=90)
plt.show()

df["grupo"] = pd.qcut(df["coluna2"], 4, labels=["baixo","medio","alto","muito_alto"])

sns.boxplot(x="grupo", y="coluna2", data=df)
plt.show()
```

```python
# 18 Substituir numa coluna respeitante a uma variável de escala,
# um determinado valor por NaN.
df["coluna2"] = df["coluna2"].replace(10050, np.nan)
```

```python
# 19 obter o número de "missing values" presentes na coluna em causa.
missing_data = df.isnull()
print(missing_data["coluna2"].value_counts())
```

```python
# 20 substituir os “missing values” presentes na coluna em causa, pela média da mesma.
media = df["coluna2"].mean()
df["coluna2"] = df["coluna2"].replace(np.nan, media)
```

```python
# 22 eliminar os “missing values” da coluna em causa.
df.dropna(subset=["coluna2"], axis=0, inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
# 24 obtero tipo de dados de cada coluna
df.dtypes, df.info()
```

```python
# 25 converter a coluna34 para datetime
df["coluna34"] = pd.to_datetime(df["coluna34"], errors="coerce")
print(df["coluna34"].dtype)
```

```python
# 26 normalizar os valores da coluna2 para o intervalo entre 0 e 1 colocando numa nova coluna
df["coluna2_norm"] = df["coluna2"] / df["coluna2"].max()
print(df["coluna2_norm"].head())
```

```python
# 27 criar uma nova coluna resultante da divisão da coluna2 com a escolha de uma variavel de escolha dividida em 3 categorias
bins = np.linspace(df["coluna2"].min(), df["coluna2"].max(), 4)
group_names = ["Baixa", "Media", "Elevada"]
df["coluna2_cat"] = pd.cut(df["coluna2"], bins=bins, labels=group_names, include_lowest=True)
print(df[["coluna2", "coluna2_cat"]].head(10))
```

```python
# 28 vamos comparar a coluna que criamos anteriormente com a que a originou 
df[['coluna2_cat','coluna2']].head(5)
```

```python
# ob534 o coeficiente de correlação entre as diferentes colunas numericas do dataset
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
```

```python
# 30 scatterplot das duas features mais fortemente relacionadas

corr = df.corr(numeric_only=True)
corr_unstacked = corr.abs().unstack()
corr_unstacked = corr_unstacked[corr_unstacked < 1]
top_pair = corr_unstacked.idxmax()
print("Par mais correlacionado:", top_pair)

col_x, col_y = top_pair

sns.regplot(x=col_x, y=col_y, data=df)
plt.xlabel(col_x)
plt.ylabel(col_y)
plt.title(f"Scatterplot entre {col_x} e {col_y}")
plt.show()
```

```python
# 31 pivot table cosiderando uma coluna a escolha, e outras duas variaveis de escala a escolha

df_new = df[["coluna29", "coluna30", "coluna2"]]
grouped = df_new.groupby(["coluna29", "coluna30"], as_index=False).mean()
pivot = grouped.pivot(index="coluna29", columns="coluna30", values="coluna2")

print(pivot)
```

```python
# 32 Apresentar gráficamente a  pivot table anterior

df_new = df[["coluna29", "coluna30", "coluna2"]]
grouped = df_new.groupby(["coluna29", "coluna30"], as_index=False).mean()
grouped_pivot = grouped.pivot(index="coluna29", columns="coluna30", values="coluna2")

plt.pcolor(grouped_pivot, cmap='RdBu')
plt.colorbar()
plt.xticks(rotation=90)
plt.show()
```

```python
# 33 apresentar a media da variável de escala anterior para cada grupo de umas das variaveis categoricas anteriores
df_groups = df[["coluna29", "coluna2"]]
grouped_by_mean = df_groups.groupby("coluna29", as_index=False).mean()

print(grouped_by_mean)
```

```python
# 34 este one-way ANOVA para os grupos de uma das variáveis categóricas anteriores

categorias = df["coluna29"].unique()
grupos = [df[df["coluna29"] == cat]["coluna2"] for cat in categorias]
f_val, p_val = stats.f_oneway(*grupos)

print("F =", f_val)
print("p =", p_val)
```
